# Steam Game Reviews & Popularity Analysis
## Fase 1 

**Onderwerp:** Hoe kan 'n speletjie-ontwikkelaar hul speletjie in 'n spesifieke genre suksesvol maak op Steam, gebaseer op resensies, spelersdata en historiese tendense?

**Datastel:** 33+ games oor 9 genres, ~65 000 skoon resensies

**Databronne:** Steam API, SteamCharts, SteamDB, ISteamNews API

**Tegnologieë:** Python 3.14, pandas, numpy, matplotlib, seaborn, scikit-learn, vaderSentiment, wordcloud, pyarrow, streamlit

**Tydperk data:** Julie 2025 – Julie 2026 (resensies), 2012–2026 (SteamDB), 2015–2026 (SteamCharts)

---
## 1. Inleiding

### Agtergrond en Konteks
Steam is die grootste PC-speletjieplatform met meer as 132 miljoen maandelikse aktiewe gebruikers en meer as 50 000 speletjies. Die platform is hoogs mededingend — na raming misluk 90% van speletjies om 'n beduidende spelersbasis te lok. Vir 'n nuwe speletjie-ontwikkelaar (veral indie) is die keuse van genre een van die mees deurslaggewende besluite wat die sukses of mislukking van 'n speletjie kan bepaal. Hierdie projek ondersoek die faktore wat bydra tot speletjie-sukses op Steam deur 'n omvattende analise van resensies, spelersdata en historiese tendense oor speletjies en 9 genres heen.

### Relevansie vir Belanghebbendes
Die projek se insigte is relevant vir verskeie belanghebbendes:
- **Speletjie-ontwikkelaars (veral indie):** Verstaan watter faktore positiewe resensies in hul teikengenre aandryf.
- **Uitgewers:** Neem ingeligte besluite oor watter genre om in te belê.
- **Beleggers:** Evalueer die potensiële sukses van speletjieprojekte gebaseer op genre-spesifieke maatstawwe.
- **Navorsers:** Verstaan die verband tussen speletjiekenmerke, spelertevredenheid en langtermynsukses.

### Monetiseringspotensiaal
Deur die faktore wat sukses in elke genre dryf te kwantifiseer, kan ontwikkelaars hul kanse op 'n positiewe ontvangs (en dus hoër verkope en spelersbehoud) aansienlik verbeter. Genre-keuse beïnvloed direk:
- Spelersbehoud en gemiddelde speeltyd
- Resensiescore en sigbaarheid op Steam
- Die potensiaal vir 'n 'verlossingsboog' deur opdaterings en gemeenskapsbestuur

### Etiese Oorwegings
Hierdie projek gebruik slegs openbare data wat beskikbaar is via Steam se amptelike API en publieke derdepartybronne (SteamCharts, SteamDB). Geen privaatgebruikersinligting word versamel of gestoor nie. Resensie-data word op 'n geanonimiseerde wyse hanteer (slegs resensieteks en metadata wat die gebruiker reeds publiek gemaak het). Daar word geen poging aangewend om gebruikersgedrag te manipuleer nie, en die projek se doel is bloot om insigte te verskaf wat ontwikkelaars kan help om beter speletjies te bou.

### Haalbaarheid
Die databronne is vrylik toeganklik:
- **Steam API:** Geen beperking op openbare resensies of speletjiebesonderhede nie.
- **SteamCharts:** Historiese spelersdata is publiek beskikbaar.
- **SteamDB:** Daaglikse resensietellings is publiek beskikbaar.
- **Tegnieke:** Alle tegnieke (NLP, regressie, tydreeksanalise) is binne die bestek van die kursus.

## 2. Probleemstelling

### Navorsingsvraag
**Hoe kan 'n speletjie-ontwikkelaar hul speletjie in 'n spesifieke genre suksesvol maak op Steam?**

### Omvang
Die projek analiseer **speletjies oor 9 genres** (RPG, Shooter, Action, Strategy, Adventure, Survival, Battle Royale, Free-to-Play, Hero Shooter). Die data dek hoofsaaklik Julie 2025 tot Julie 2026, maar SteamDB-data strek terug tot die bekendstellingsdatum van elke speletjie (sommige tot 2012).

### Hipotese
**Genre bepaal die basislyn-verwagtinge van spelers.** Suksesfaktore soos speeltyd, resensielengte, opdateringsfrekwensie en sentiment verskil beduidend tussen genres. 'n Ontwikkelaar moet hul mededingers binne hul teikengenre meet, nie die hele speletjiemark nie. Byvoorbeeld, 'n 70% positiewe resensiescore kan uitstekend wees vir 'n Free-to-Play-speletjie, maar teleurstellend vir 'n Action-RPG.

### Sub-Navorsingsvrae
1. **Watter kenmerke voorspel 'n positiewe resensie?** Speeltyd, resensielengte, stemme, aankoop-type?
2. **Verskil suksesfaktore per genre?** Is speeltyd belangriker vir RPG's as vir Shooters?
3. **Hoe ontwikkel spelertevredenheid oor tyd?** Kan 'n 'verlossingsboog' (soos Cyberpunk 2077) voorspel word?
4. **Watter rol speel inhoudopdaterings?** Het speletjies wat gereeld opdateer 'n hoër resensiescore?
5. **Wat is die tipiese seisoenspatrone?** Is daar beter/seisoene vir speletjie-bekendstellings?

### Uitdagings
- **Steam API-perke:** Die API keer slegs ~5 000 resensies per speletjie terug in 'n enkele sessie (alhoewel daar maniere is om meer te kry maar vat te lank).
- **Taalvariëteit:** Resensies is in 29+ tale; VADER werk die beste met Engels.
- **SteamDB-kumulatiewe data:** Sommige speletjies (soos Overwatch 2) se eerste datapunt is 'n kumulatiewe totaal, nie 'n daaglikse delta nie. 
- **Resensie-bomaanvalle:** byvoorbeeld Helldivers 2 se PSN-koppeling-kontroversie het enorme daaglikse resensiepieke veroorsaak. Die ratio-skatting is aangepas om nie kunsmatige data te skep tydens onstabiele periodes nie.

### Potensiële Impak
Hierdie projek kwantifiseer die faktore wat speletjie-sukses op Steam dryf, wat ontwikkelaars in staat stel om:
- Data-gedrewe besluite te neem oor genre-keuse
- Te verstaan watter kenmerke om te prioritiseer tydens ontwikkeling
- Realistiese verwagtinge te stel vir resensiescore gebaseer op hul genre
- Tydsberekening van bekendstellings en opdaterings te optimaliseer

## 3. Metodologie

### 3.1 Data-Insameling

**Steam API (src/scrape.py):**
- **Resensies:** `ISteamReviews/GetAppReviews` — `purchase_type=all`, `day_range=9999`, tot 100 bladsye, 0.3s vertraging, deduplikasie via `seen_ids`, stop na 3 leë bladsye.
- **Speletjiebesonderhede:** `ISteamApps/GetAppList` + individuele app-aanvrae vir kategorieë, genres, ondersteuning.
- **Spelersgetalle:** `ISteamUserStats/GetNumberOfCurrentPlayers` — tans aktiewe spelers vir al 33 speletjies.
- **Inhoudgebeure:** `ISteamNews/GetNewsForApp` — keyword-filters: patch, update, dlc, expansion, season, event, content, hotfix. 836 gebeure gestoor.

**SteamCharts (src/steamcharts.py):**
- Historiese maandelikse gemiddelde en piek-spelersdata (705 rye, 33 speletjies).
- Datumontleding: `pd.to_datetime('01 ' + df['month'], format='%d %B %Y')`.

**SteamDB (src/steamdb.py):**
- Daaglikse positiewe en negatiewe resensietellings, gestoor as `data/raw/steamdb_chart_{app_id}.csv` (33 lêers).
- Data word gesuiwer met `_remove_cumulative_first_row()` (bespeur en verwyder kumulatiewe eerste rye) en `_estimate_ratio()` (vul ontbrekende data in, maar slaan skatting oor tydens onstabiele periodes soos resensie-bomaanvalle).
- Maandeliks geaggregeerd (2831 rye, 33 speletjies) gestoor as `data/processed/steamdb_monthly.csv`.

### 3.2 Data-Verwerking (src/clean.py)

Die skoonmaakproses sluit in:
- **Deduplikasie:** Verwyder duplikate gebaseer op `review_id`.
- **Teksskoonmaak:** Verwyder URL's, nie-alfabetiese karakters.
- **Lengtefilter:** Verwyder resensies korter as 10 karakters.
- **Speeltyd-uiterstes:** Knip ekstreme uitskieters by die 99ste persentiel.
- **Kenmerkingenieurswese:** `review_date`, `word_count`, `review_length`, `has_early_access`, `is_steam_purchase`.
- **Genre-one-hot:** Nege genre-aanwyserkolomme (`genre_RPG`, `genre_Shooter`, ens.) via `GAME_GENRES`-kartering.
- **Resultaat:** ~65 000 skoon resensies (van ~66 000 rou).

### 3.3 NLP-Analise (src/nlp_analysis.py)

- **VADER-sentiment:** Vyf kolomme — `vader_compound`, `vader_positive`, `vader_neutral`, `vader_negative`, `vader_sentiment_label`. Vooraf bereken en in CSV gestoor.
- **TF-IDF:** Per speletjie (top 20 terme) en per genre, met `ngram_range=(1,1)` om OOM-foute op WSL te vermy. Steekproef tot 30 000 resensies indien nodig.
- **Woordwolke:** Positiewe en negatiewe resensies, tot 500 monsters per wolk.

### 3.4 Regressie-Analise (src/regression.py)

- **Lineêre regressie:** Voorspel `vader_compound` gebaseer op speeltyd, resensielengte, woordtelling, genre-aanwysers, per-speletjie-aanwysers. Sluit R² en RMSE in.
- **Logistiese regressie:** Voorspel of `voted_up` waar is. Sluit AUC-ROC en klassifikasieverslag in.
- **Kenmerke:** Beide modelle gebruik dieselfde voorbereide kenmerke, met 'n klein verskil: logistiese regressie voeg `vader_compound` by as 'n kenmerk.

### 3.5 Tydreeksanalise (src/timeseries.py)

- **Daaglikse resensievolume:** Daaglikse telling van resensies.
- **Maandelikse aggregasie:** Totale resensies en positiewe persentasie per maand.
- **Verlossingsboë:** Vergelyk eerste 3 maande se positiewe % met laaste 3 maande. Beide resensie- en SteamDB-weergawes.
- **Seisoenpatrone:** Resensies per uur, dag van week, maand.
- **Genre-tendense:** Maandelikse positiewe % per genre, met inhoudgebeure-merkers.
- **SteamDB-aanvulling:** Historiese daaglikse data van SteamDB vir akkurate volume- en tendensanalise.

### 3.6 Dashboard (dashboard/app.py)

Interaktiewe Streamlit-dashboard met 5 tabs:
- **Oorsig:** Totale resensies, positiewe %, per-speletjie-tabelle, per-genre-tabelle
- **Spelers & Tyd:** Daaglikse volume (SteamDB), maandelikse % (SteamDB), seisoenspatrone, huidige spelers, SteamCharts-geskiedenis, genre-tendense, inhoudgebeure
- **NLP:** VADER-verspreiding, woordwolke, TF-IDF per speletjie en per genre
- **Vergelyk:** Speletjie- of genre-vergelyking, verstrooiingsdiagramme, verlossingsboë
- **Voorspellings:** Lineêre en logistiese regressie-uitsette

**Optimisering vir WSL:** `@st.cache_resource`, `string_storage='python'`, `runOnSave=false`, `faulthandler.enable()`, `pyarrow<25`.

## 4. Fase 1 Opsomming

In Fase 1 is die volgende bereik:

1. **Probleem gedefinieer:** Duidelike navorsingsvraag oor hoe speletjie-ontwikkelaars sukses op Steam kan behaal, met genre as die primêre konteks.

2. **Hipotese geformuleer:** Genre bepaal basislyn-verwagtinge; suksesfaktore verskil tussen genres.

3. **Databronne geïdentifiseer:** Steam API, SteamCharts, SteamDB, ISteamNews — almal publiek en vrylik toeganklik.

4. **Metodologie beplan en geïmplementeer:**
   - Data-insameling via web scraping (33 speletjies, ~66k resensies)
   - Data-verwerking en skoonmaak (pandas, ~65k skoon resensies)
   - NLP-analise (VADER, TF-IDF, woordwolke)
   - Regressiemodelle (lineêr en logisties)
   - Tydreeksanalise (volume, tendense, verlossingsboë, seisoene)
   - Interaktiewe Streamlit-dashboard

5. **Uitdagings aangespreek:**
   - PyArrow WSL2-segfault (pyarrow<24, string_storage='python')
   - SteamDB-kumulatiewe data (_remove_cumulative_first_row)
   - Resensie-bomaanvalle (ratio_std_threshold in _estimate_ratio)
   - OOM in TF-IDF en WordCloud (steekproefneming)

6. **Etiese aspekte gedokumenteer:** Slegs openbare data, geen PII, geen manipulasie.
